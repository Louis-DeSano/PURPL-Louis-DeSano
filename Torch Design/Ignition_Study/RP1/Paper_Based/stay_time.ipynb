```python
import numpy as np
```

```python
# Propellant Densities [lbm/in^3]
rho_ox = 0.0412
rho_kero = 0.035
rho_avg = (rho_ox + rho_kero) / 2

#MCA Set Points
mdot_main = 20 #[lbm/s]
Lstar = 45 #[in]
d_throat = 2.99 #[in]
A_t = np.pi * (d_throat/2)**2 #[in^2]

V_c = Lstar * A_t
```

```python
t_stay = V_c * rho_avg / mdot_main #[s]
t_stay
```
