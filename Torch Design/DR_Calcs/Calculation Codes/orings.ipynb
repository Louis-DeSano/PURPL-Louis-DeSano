# Get Compressive Load Force Required For O-Rings Used

### O-Ring Parameters

```python
# General -100 Series
width = 0.103 #[in]
durometer = "75A"
squeeze = [17, 24] #Squeeze Percent

# Specific Dimensions [in]
ID_122 = [1.318]
OD_122 = 0.755
ID_117 = [0.799 - 0.010, 0.799 + 0.010]
OD_117 = 1.005

mid_diam_122 = np.mean([np.mean(ID_122), OD_122])
mid_diam_117 = np.mean([np.mean(ID_117), OD_117])

mid_circ_122 = np.pi * mid_diam_122
mid_circ_117 = np.pi * mid_diam_117

mid_circ_set = [mid_circ_122, mid_circ_117]
dash_set = [122, 117]
mid_circ_df = pd.DataFrame([mid_circ_set], columns=[dash_set], index=["Midline Circumference [in]"])
mid_circ_df
```
